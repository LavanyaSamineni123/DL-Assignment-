```python
# Loading model and tokenizer with Hugging Face token authentication
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline
```

```python
# Included Replaced Hugging Face token
huggingface_token = "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"

# Load the tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("facebook/bart-large-mnli", use_auth_token=huggingface_token)
model = AutoModelForSequenceClassification.from_pretrained("facebook/bart-large-mnli", use_auth_token=huggingface_token)

# Initializing the zero-shot classifier
zero_shot_classifier = pipeline("zero-shot-classification", model=model, tokenizer=tokenizer)

# Defining candidate labels
candidate_labels = ["scientific", "non-scientific"]
```

```python
#data

texts = [
    "The study of quantum mechanics reveals the fundamental nature of particles.",
    "I love going to the beach and soaking up the sun.",
    "Theoretical physics is a branch of physics that employs mathematical models.",
    "Pizza is my favorite food, especially with extra cheese."
]
```

```python
# Compute scores for each text
for text in texts:
    result = zero_shot_classifier(text, candidate_labels)
    scientific_score = result['scores'][result['labels'].index('scientific')]
    non_scientific_score = result['scores'][result['labels'].index('non-scientific')]
    print(f"Text: {text}")
    print(f"Scientific Score (Zero-Shot): {scientific_score:.4f}")
    print(f"Non-Scientific Score (Zero-Shot): {non_scientific_score:.4f}")
    print("-" * 40)
```

BERT CLASSIFIER:

```python
# Load the tokenizer and model
model_name = "allenai/scibert_scivocab_uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name, use_auth_token=huggingface_token)
model = AutoModelForSequenceClassification.from_pretrained(model_name, use_auth_token=huggingface_token)
scientific_classifier = pipeline("text-classification", model="allenai/scibert_scivocab_uncased")
```

```python
# Example text
texts = [
    "The study of quantum mechanics reveals the fundamental nature of particles.",
    "I love going to the beach and soaking up the sun.",
    "Theoretical physics is a branch of physics that employs mathematical models.",
    "Pizza is my favorite food, especially with extra cheese."
]
```

```python
# Compute scores for each text
for text in texts:
    result = scientific_classifier(text)  # Pass individual text, not the entire list
    scientific_score = result[0]['score']  # Extract the score for the first label
    non_scientific_score = 1 - scientific_score  # Non-scientific score is the complement of scientific score
    print(f"Text: {text}")
    print(f"Scientific Score (BERT): {scientific_score:.4f}")
    print(f"Non-Scientific Score (BERT): {non_scientific_score:.4f}")
    print("-" * 40)
```

SUMMARY:

Zero_Short_Model:
Gave High Scores for scientific texts: The zero-shot classifier assigns high scores to texts that are explicitly scientific, such as "The study of quantum mechanics reveals the fundamental nature of particles" and "Theoretical physics is a branch of physics that employs mathematical models." This indicates that the zero-shot model is effective at identifying scientific content without being explicitly trained on a labeled dataset.

Gave Low Scores for the texts which are not scientific: The zero-shot classifier assigns low scores to non-scientific texts, such as "I love going to the beach and soaking up the sun" and "Pizza is my favorite food, especially with extra cheese." This shows that the model can distinguish between scientific and non-scientific content.

BERT MODEL:
Bert model gave almost similar score near to 0.5 for all the given texts, irrespective of their scientific nature.This indicates that the BERT model is a nit less effective at distinguishing between scientific and non-scientific content in this context. and In my opinion it has to be fine tuned further more in distinguishing the scientific and non scientific text.
